```python
import torch
import torchvision

print("Torch:", torch.__version__)
print("Torchvision:", torchvision.__version__)

print("\nCUDA available:", torch.cuda.is_available())

# Test torchvision NMS
boxes = torch.tensor([
    [10, 10, 100, 100],
    [20, 20, 90, 90]
])

scores = torch.tensor([0.9, 0.8])

print("\nTesting torchvision NMS...")

try:
    result = torchvision.ops.nms(boxes, scores, 0.5)
    print("NMS CPU test: SUCCESS")
    print(result)
except Exception as e:
    print("NMS CPU test FAILED:")
    print(type(e).__name__)
    print(e)


if torch.cuda.is_available():

    boxes_cuda = boxes.cuda()
    scores_cuda = scores.cuda()

    print("\nTesting CUDA NMS...")

    try:
        result = torchvision.ops.nms(
            boxes_cuda,
            scores_cuda,
            0.5
        )

        print("NMS CUDA test: SUCCESS")
        print(result)

    except Exception as e:
        print("NMS CUDA test FAILED:")
        print(type(e).__name__)
        print(e)
```
